
#### 1.3  Set PassengerId as the index

#### 1.4 Create a pie chart presenting the male/female proportion

* for all passangers
* for surviving passengers

What is your conclusion?


```python
#of the all dataset
men_all=df[df["Sex"]=='male'].shape[0]
women_all=df[df["Sex"]=='female'].shape[0]


#of the numer of suviving passengers
survived_male = df[(df['Survived'] == 1) & (df['Sex'] == 'male')]
survived_female = df[(df['Survived'] == 1) & (df['Sex'] == 'female')]


dem=df.shape
w_p=survived_female.shape[0]*100/dem[0]
m_p=100-w_p

labels=["Male","Female"]

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].pie([men_all,women_all], labels=["Men","Women"], colors=["green","red"], autopct='%1.1f%%', startangle=140)
axes[0].set_title('The pie chart presenting the male/female proportion for all passangers')

axes[1].pie([m_p,w_p], labels=["Men","Women"], colors=["green","red"], autopct='%1.1f%%', startangle=140)
axes[1].set_title('The pie chart presenting the male/female proportion for surviving passengers')

#for round shape
plt.axis('equal')
plt.show()
```

#### 1.5. Create a scatterplot with the Fare payed and the Age, differ the plot color by gender

```python
fig=plt.figure()
ax=fig.add_subplot(1,1,1)
colors = {'male': 'blue', 'female': 'pink'}
plt.scatter(df["Age"], df["Fare"], c=[colors[g] for g in df["Sex"]], label=df["Sex"])

plt.xlabel('Age')
plt.ylabel('Fare Paid')
plt.legend(loc='upper right')

legend_labels = {'male': 'blue circles - men', 'female': 'рink circles - women'}
legend_elements = [plt.Line2D([0], [0], marker='o', color='w', markerfacecolor=colors[key], markersize=8, label=legend_labels[key]) for key in colors.keys()]
ax.legend(handles=legend_elements, loc='upper right')




# Show the plot
plt.title('Scatterplot of Fare Paid vs. Age by Gender')
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder


df=pd.read_csv('train.csv',delimiter=',')
pd.read_csv('train.csv',delimiter=',')[:5]
```

```python

# Просмотрев датасет можно заметить, что существуют столбцы в которых почти каждое значение равняется NaN

# Проверяем верна ли наша теория
nan_count = df['PoolQC'].isna().sum()
print(df.shape)
print(nan_count)


# Для проверки остальных столбцов построим диаграмму для наглядности
fig = plt.figure(figsize=(20, 25))

# Подсчёт количество NaN значений в каждом столбце
nan_counts = df.isna().sum()


nan_counts.plot(kind='barh')  # Используем barh для горизонтального столбчатого графика
plt.title('Количество NaN значений по столбцам')
plt.ylabel('Столбцы')
plt.xlabel('Количество NaN')

# Устанавливаем вертикальные подписи снизу по оси y
plt.xticks(rotation=0)
plt.show()

# Из графика видим, что и вправду существует несколько столбцов значения NaN в которых близки к размеру самого датасета
# С такими столбцами работать нет смысла и поэтому удаляем их




# Параллельно буду преобразовывать тестовые данные
# Подготовка тестовых данных
#1)Объединю test.csv и sample_submission.csv, чтобы при дальнейшей подготовке данных не нужно было тратить время на стравниванияе id

df_1=pd.read_csv('test.csv',delimiter=',')
df_2=pd.read_csv('sample_submission.csv',delimiter=',')
df_2= df_2.drop('Id', axis=1)
full_test_df=pd.concat([df_1, df_2], axis=1)


# Создадим список названий столбцов, где количество NaN больше 1000 и удалим все такие столбцы из датафрэйма
more_than_1000_nan = nan_counts[nan_counts > 1000].index.tolist()
df = df.drop(columns=more_than_1000_nan)
full_test_df= full_test_df.drop(columns=more_than_1000_nan)

# Построим новую диаграмму и посмотрим что мы имеем на данный момент
fig = plt.figure(figsize=(20, 25))
nan_counts = df.isna().sum()
nan_counts.plot(kind='barh')  # Используем barh для горизонтального столбчатого графика
plt.title('Количество NaN значений по столбцам ')
plt.ylabel('Столбцы')
plt.xlabel('Количество NaN')
plt.show()


# Видим, что ситуация сильно улучшилась и остался лишь 1 столбец у которого значений NaN примерно половина от всего датасета. Его удалим вручную
df = df.drop(columns='FireplaceQu')
full_test_df=full_test_df.drop(columns='FireplaceQu')


# Посмотрим можно ли удалить все строки, где встречается NaN.
print(df.shape)
df_nan = df.dropna()
print(df_nan.shape)

#Видим, что удаляется примерно 1/3 датасета, поэтому просто удалять строки с NaN не является лучшим вариантом
#Поэтому заполним значения NaN с помощью натболее подходящего для данного случая метода -метод заполнения NaN средним значением этого столбца

# Заполните все NaN значения в датасете средними значениями для каждого ЧИСЛОВОГО столбца
df = df.fillna(df.mean())
full_test_df=full_test_df.fillna(df.mean())

print(df.shape)
print(full_test_df.shape)

```

```python
#Проверим остались ли NaN, построив круговую диаграмму

# Определим, в каких столбцах есть хотя бы одно NaN значение
columns_with_nan = df.columns[df.isna().any()].tolist()

# Вычислите процент столбцов с NaN значениями
percentage_with_nan = (len(columns_with_nan) / len(df.columns)) * 100

# Создадим круговую диаграмму
plt.figure(figsize=(6, 6))
plt.pie([percentage_with_nan, 100 - percentage_with_nan], labels=["Contains NaN", "Without NaN"], autopct='%1.1f%%')
plt.title('Percentage of columns with NaN values')
plt.show()

# Видим, что в датасете всё же есть NaN в категориальныех значениях
# Построим диаграмму, чтобы оценить их объём

fig = plt.figure(figsize=(20, 25))
nan_counts = df.isna().sum()
nan_counts.plot(kind='barh')  # Используем barh для горизонтального столбчатого графика
plt.title('Number of NaN values by columns')
plt.ylabel('Columns')
plt.xlabel('Number of NaN')
plt.show()

# Видим, что NaN значений осталось мало, поэтому теперь можно удалить строки, где они содержаться без вреда для датасета
df = df.dropna()

print(df.shape)
print(full_test_df.shape)

```

```python
# Построим круг. диаграмму для проверки нет ли NaN в датасете

# Определим, в каких столбцах есть хотя бы одно NaN значение
columns_with_nan = df.columns[df.isna().any()].tolist()

# Вычислите процент столбцов с NaN значениями
percentage_with_nan = (len(columns_with_nan) / len(df.columns)) * 100

# Создадим круговую диаграмму
plt.figure(figsize=(6, 6))
plt.pie([percentage_with_nan, 100 - percentage_with_nan], labels=["Contains NaN", "Without NaN"], autopct='%1.1f%%')
plt.title('Percentage of columns with NaN values')
plt.show()

# Изучив диаграмму, видим, что NaN значений больше нет и всё было сделано верно
```

```python

# Определям типы признаков числовые и категориальные
numeric_features = df.select_dtypes(include=['int', 'float']).columns
categorical_features = df.select_dtypes(include=['object']).columns

#Отделим категориальные данные чтобы сразу через ohe прогнать
categ_df=df[categorical_features]

columns_with_nan = categ_df.columns[categ_df.isna().any()].tolist()
# Вычислите процент столбцов с NaN значениями
percentage_with_nan = (len(columns_with_nan) / len(numeric_features)) * 100
plt.figure(figsize=(6, 6))
plt.pie([percentage_with_nan, 100 - percentage_with_nan], labels=["Contains NaN", "Without NaN"], autopct='%1.1f%%')
plt.title('Percentage of columns with NaN values in categ_df')
plt.show()

numeric_df = df[numeric_features]


# Определите, в каких числовых столбцах есть хотя бы одно NaN значение
columns_with_nan = numeric_df.columns[numeric_df.isna().any()].tolist()
# Вычислите процент столбцов с NaN значениями
percentage_with_nan = (len(columns_with_nan) / len(numeric_features)) * 100


plt.figure(figsize=(6, 6))
plt.pie([percentage_with_nan, 100 - percentage_with_nan], labels=["Contains NaN", "Without NaN"], autopct='%1.1f%%')
plt.title('Percentage of columns with NaN values in numeric_df')
plt.show()

# Теперь у вас есть списки числовых и категориальных признаков
print("Numerical signs:", numeric_features)
print("Categorical features:", categorical_features)

print(df.shape)
print(full_test_df.shape)
```

```python

from sklearn.preprocessing import OneHotEncoder# creating instance of one-hot-encoder
# Создание объекта ohe
encoder = OneHotEncoder(sparse=False)  # sparse=False для получения плотной матрицы
ohe_data = encoder.fit_transform(categ_df)
#print(type(ohe_data))

# Преобразование ohe_data в DataFrame
ohe_df = pd.DataFrame(ohe_data, columns=encoder.get_feature_names_out(input_features=categorical_features))
#объединение категориальных и числовых данных
full_df=pd.concat([ohe_df,numeric_df], axis=1)
full_df= full_df.dropna()
print(full_df.shape)
```

```python
# Определям типы признаков числовые и категориальные
numeric_features = full_test_df.select_dtypes(include=['int', 'float']).columns
categorical_features = full_test_df.select_dtypes(include=['object']).columns

#Отделим категориальные данные чтобы сразу через ohe прогнать
categ_df=full_test_df[categorical_features]

numeric_df = full_test_df[numeric_features]

from sklearn.preprocessing import OneHotEncoder# creating instance of one-hot-encoder
# Создание объекта ohe
encoder_1 = OneHotEncoder(sparse=False)  # sparse=False для получения плотной матрицы
ohe_data = encoder_1.fit_transform(categ_df)
#print(type(ohe_data))

# Преобразование ohe_data в DataFrame
ohe_df = pd.DataFrame(ohe_data, columns=encoder_1.get_feature_names_out(input_features=categorical_features))
#объединение категориальных и числовых данных
full_df_testovoe=pd.concat([ohe_df,numeric_df], axis=1)


print(full_df.shape)
print(full_df_testovoe.shape)
```
