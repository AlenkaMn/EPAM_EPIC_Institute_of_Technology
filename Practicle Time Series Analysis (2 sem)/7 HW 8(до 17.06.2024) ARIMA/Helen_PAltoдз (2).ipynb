#D. Find the order of integration

Run the ADF test and check if the process is stationary. If not, apply differencing.


I will use Augmented Dickey–Fuller (ADF) Test to check if the data is stationary.

The null hypothesis: there is a unit root present in a time series.

The augmented Dickey-Fuller (ADF) test helps us determine if a time series is stationary by testing for the presence of a unit root. If a unit root is present, the time series is not stationary.

```python
from statsmodels.tsa.stattools import adfuller


ADF_result = adfuller(new_df['Millions'])
print(f'ADF Statistic: {ADF_result[0]}')
print(f'p-value: {ADF_result[1]}')
```

An ADF statistic of -2.654639251341716 and a p-value of  0.08222189585453338. The ADF statistic is not negative number, and with a p-value greater than 0.05, we can not reject the null hypothesis.the series is not stationary.

Conclusion: the dataset is **not** stationary.

```python
from typing import Union
import pandas as pd
from tqdm import tqdm
from statsmodels.tsa.statespace.sarimax import SARIMAX

def optimize_SARIMA(endog: Union[pd.Series, list], order_list: list, d: int) -> pd.DataFrame:
    results = []
    # Итерация по списку параметров модели SARIMA
    for order in tqdm(order_list, desc='Fitting models'):
        try:
            # Создание и обучение модели SARIMA
            model = SARIMAX(endog, order=(order[0], d, order[1]), simple_differencing=False).fit(disp=False)
            # Получение значения AIC (критерий информационного критерия Акаике)
            aic = model.aic
            # Сохранение результатов в формате [параметры модели, AIC]
            results.append([order, aic])
        except Exception as e:
            print(f"An error occurred for order {order}: {e}")
            continue

    # Создание DataFrame из результатов
    result_df = pd.DataFrame(results, columns=['(p, q)', 'AIC'])
    # Сортировка по возрастанию значения AIC (меньше AIC - лучше модель)
    result_df = result_df.sort_values(by='AIC', ascending=True).reset_index(drop=True)

    return result_df
```

```python
d = 2

from itertools import product
ps = range(0, 13, 1)
qs = range(0, 13, 1)
```

```python
ARIMA_order_list = list(product(ps, qs))
```

```python
ARIMA_statistic=optimize_SARIMA(train,ARIMA_order_list, d)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.stats.diagnostic import acorr_ljungbox

# Получение остатков модели
residuals = ARIMA_model_fit.resid

# Q-Q plot
plt.figure(figsize=(10, 6))
sm.qqplot(residuals, line='s')
plt.title('Q-Q Plot of Residuals')
plt.show()

# Гистограмма остатков
plt.figure(figsize=(10, 6))
sns.histplot(residuals, kde=True)
plt.title('Histogram of Residuals')
plt.show()

# ACF и PACF остатков
fig, ax = plt.subplots(1, 2, figsize=(16, 6))
plot_acf(residuals, lags=40, ax=ax[0])
plot_pacf(residuals, lags=40, ax=ax[1])
plt.show()

# Тест Льюнга-Бокса для автокорреляции остатков
ljung_box_results = acorr_ljungbox(residuals, lags=[10], return_df=True)
print(ljung_box_results)
```

```python
def rolling_forecast(df: np.ndarray, train_len: int, horizon: int, window: int, method: str, p: int, d:int,q: int) -> list:
    total_len = train_len + horizon
    pred_values = []
    if method == 'mean':
        pred_mean = []
        for i in range(train_len, total_len, window):
            mean = np.mean(df[:i])
            pred_mean.extend(mean for _ in range(window))
        return pred_mean


    elif method == 'ARIMA':
        pred_ARMA = []
        for i in range(train_len, total_len, window):
            model = SARIMAX(df[:i], order=(p, d, q))
            res = model.fit(disp=False)
            predictions = res.get_prediction(0, i + window - 1)
            pred = predictions.predicted_mean[-window:]  # Использование среза NumPy
            pred_ARMA.extend(pred)
        return pred_ARMA

    return pred_values
```

```python
horizon = 53 # Горизонт прогнозирования
window = 2  # Длина окна
train_len=len(train)
print(len(test))

pred_values_mean = rolling_forecast(train, len(train), horizon, window, 'mean',p=1,q=1,d=2)
pred_values_ARIMA = rolling_forecast(train, len(train), horizon, window, 'ARIMA',p=11,q=2,d=2)
```
